```python
from numpy import diff
from ipywidgets import widgets
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display, clear_output

url = "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?query=select%20%2A%20from%20pscomppars&format=csv"

df = pd.read_csv(url)
cols =['pl_name', 'hostname', 'pl_rade', 'pl_orbper', 'pl_eqt', 'discoverymethod']
clean_df = df[cols].drop_duplicates(subset=['pl_name']).dropna(subset=['pl_name'])
print("Data Successfully Loaded!")

```

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import pandas as pd
def show_planet(p):
  d = clean_df[clean_df['pl_name'] == p].iloc[0]
  r = d['pl_rade']
  p = d['pl_name']
  avg = clean_df['pl_rade'].mean()
  print("Planet:", d['pl_name'])
  print("Host Star:", d['hostname'])
  print("Radius:", d['pl_rade'], "Earth radii")
  print("Orbital Period:", d['pl_orbper'], "days")
  print("Temperature:", d['pl_eqt'], "K")
  print("Discovering Method:", d['discoverymethod'])
  print("-" * 30)

  plt.figure(figsize=(6,3))
  plt.bar(['Selected Planet', 'Dataset Average'], [r, avg])
  plt.ylabel("Radius (Earth Radii)")
  plt.title(f"Radius Comparison: {p}")
  plt.show()

widgets.interact(show_planet, p=sorted(clean_df['pl_name'].tolist()))
```

```python
import seaborn as sns

plt.figure(figsize=(10,5))
clean_df['discoverymethod'].value_counts().plot(kind='bar', color='teal')
plt.title("Number of Exoplanets Discovered by Method")
plt.xlabel("Discovery Method")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(clean_df['pl_orbper'], clean_df['pl_rade'], alpha=0.5, color='purple')
plt.xscale('log')
plt.yscale('log')
plt.xlabel("Orbital Period (days)")
plt.ylabel("Planet Radius (Earth Radii)")
plt.title("Orbital Period vs Planet Radius")
plt.show()
```
